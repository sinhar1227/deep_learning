### As there is only 3 uniques, so do one hot encoding for this column

```python
from sklearn.preprocessing import OneHotEncoder

onehot_encoder_geo = OneHotEncoder()
geo_encoder = onehot_encoder_geo.fit_transform(df[['Geography']])
geo_encoder.toarray()
```

```python
geo_features = onehot_encoder_geo.get_feature_names_out(['Geography'])
geo_features
```

```python
df_geo = pd.DataFrame(geo_encoder.toarray(), columns=geo_features)
df_geo.head()
```

```python
df_new = pd.concat([df.drop('Geography',axis=1), df_geo], axis=1)
df_new.head()
```

### Do the encoding for Gender column

```python
from sklearn.preprocessing import LabelEncoder

labelencoder_gender = LabelEncoder()
gender_encoder = labelencoder_gender.fit_transform(df_new['Gender'])
df_new['Gender'] = gender_encoder

df_new.head()
```
