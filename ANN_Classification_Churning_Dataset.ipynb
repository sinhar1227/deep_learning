### Explore the data

```python
import pandas as pd

df = pd.read_csv('/content/sample_data/Churn_Modelling.csv')
```

```python
df.head()
```

### Drop irrelevant coloumns

```python
irrelevant_columns = ['RowNumber', 'CustomerId', 'Surname']
df = df.drop(irrelevant_columns, axis=1)
```

```python
df.head()
```

### Encode non-numeric columns

```python
df['Geography'].unique()
```

### As there is only 3 uniques, so do one hot encoding for this column

```python
from sklearn.preprocessing import OneHotEncoder

onehot_encoder_geo = OneHotEncoder()
geo_encoder = onehot_encoder_geo.fit_transform(df[['Geography']])
geo_encoder.toarray()
```

```python
geo_features = onehot_encoder_geo.get_feature_names_out(['Geography'])
geo_features
```

```python
df_geo = pd.DataFrame(geo_encoder.toarray(), columns=geo_features)
df_geo.head()
```

```python
df_new = pd.concat([df.drop('Geography',axis=1), df_geo], axis=1)
df_new.head()
```

### Do the encoding for Gender column

```python
from sklearn.preprocessing import LabelEncoder

labelencoder_gender = LabelEncoder()
gender_encoder = labelencoder_gender.fit_transform(df_new['Gender'])
df_new['Gender'] = gender_encoder

df_new.head()
```

### Next step scaling the different columns, as we only need to do scaling for dependent coloumn, so first split the data into train and test

```python
from sklearn.model_selection import train_test_split

x=df_new.drop('Exited', axis=1)
y=df_new[['Exited']]

X_train, X_test, y_train, y_test = train_test_split(x,y,test_size=0.20)
```

### Let scaled the data

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.fit_transform(X_test)
```

```python
X_test_scaled
```

```python
X_test
```

```python
X_test_scaled.shape
```

### Create ANN for classification

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
```

```python
model=Sequential([
   Input((X_train_scaled.shape[1],)),
   Dense(32,activation='relu'),
   Dense(1,activation='sigmoid')
])
model.compile(loss='binary_crossentropy',optimizer='adam')
```

```python
model.fit(X_train_scaled, y_train,batch_size=5,epochs=50)
```

```python
y_train_predict=model.predict(X_train_scaled)
y_test_predict=model.predict(X_test_scaled)
```
