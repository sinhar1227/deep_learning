### BERT - Bidirectional Encoder Representations from Transformers

BERT is a transformer-based language model developed by Google for Natural Language Processing (NLP) tasks.

It introduced the idea of **bidirectional context learning**, meaning it reads text from both left-to-right and right-to-left at the same time to understand word meaning based on surrounding words.

| Model                   | Description        |
| ----------------------- | ------------------ |
| bert-base-uncased       | Lowercase English  |
| bert-large-uncased      | Larger version     |
| distilbert-base-uncased | Faster and smaller |
| roberta-base            | Improved BERT      |
| albert-base-v2          | Lightweight BERT   |

# Why BERT Was Important

Before BERT:

* Models usually read text in one direction
* Context understanding was limited

Example:

> “bank” in:

* “river bank”
* “bank account”

Older models struggled with contextual meaning.

BERT understands context using surrounding words from both directions.

---

# Architecture

BERT is based on the Transformer architecture introduced in:

Transformer

Specifically:

* Uses only the **Encoder** part of Transformer
* Built using:

  * Multi-head self-attention
  * Feed-forward neural networks
  * Positional encoding

---

# Core Idea: Bidirectional Learning

Traditional models:

```text
Left → Right
```

BERT:

```text
Left ↔ Right
```

It learns context from both sides simultaneously.

Example:

```text
The man went to the bank to withdraw money.
```

BERT understands “bank” means financial institution.

# Pretraining Tasks

## 1. Masked Language Modeling (MLM)

Some words are hidden.

Example:

```text
I love [MASK] learning.
```

BERT predicts:

```text
deep
```

---

## 2. Next Sentence Prediction (NSP)

Model learns sentence relationships.

Example:

Sentence A:

```text
I went to school.
```

Sentence B:

```text
I attended math class.
```

BERT predicts whether B follows A logically.

Load BERT Model and Tokenizer Manually

````python
from transformers import BertTokenizer, BertModel
import torch

# Load tokenizer
tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")

# Load pretrained BERT model
model = BertModel.from_pretrained("bert-base-uncased")

text = "BERT understands context very well."

# Tokenize input
inputs = tokenizer(text, return_tensors="pt")

# Generate embeddings
with torch.no_grad():
    outputs = model(**inputs)

# Last hidden states
last_hidden_states = outputs.last_hidden_state

print(last_hidden_states.shape)
````

Extract Sentence Embeddings

````python
from transformers import BertTokenizer, BertModel
import torch

tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")
model = BertModel.from_pretrained("bert-base-uncased")

sentence = "Artificial Intelligence is transforming the world."

inputs = tokenizer(sentence, return_tensors="pt")

with torch.no_grad():
    outputs = model(**inputs)

# CLS token embedding
sentence_embedding = outputs.last_hidden_state[:, 0, :]

print(sentence_embedding.shape)
````
